# **Descriptive Statistics: Diamond Dataset**

### This notebook applies:

1. Basic inspection of a dataset before any statistics are computed
2. Descriptive statistics for numerical variables: centre, spread and shape
3. Descriptive statistics for categorical variables: frequencies, shares and mode
4. Group statistics with `groupby`
5. Bivariate analysis: covariance, correlation and contingency tables

Dataset: `Diamond.csv` (Chu, 2001) - 308 observations, 5 variables.
Source of the data: https://jse.amstat.org/v9n2/datasets.chu.html

## **1. Import Libraries**

```python
# 1. Import libraries
# pandas handles the table itself; numpy is used for a single numerical helper below.

import pandas as pd
import numpy as np

# Print floating point numbers with two decimals instead of scientific notation.
# Without this line pandas prints the variance of price as 1.158120e+07.
pd.set_option("display.float_format", "{:.2f}".format)
```

## **2. Load and Inspect the Dataset**

```python
# 2. Load the dataset
# The file is expected to sit next to this notebook, so no full path is needed.

df = pd.read_csv("Diamond.csv")

# Show the first ten rows to see what the table actually looks like.
df.head(10)

# Interpretation:
# Five columns: carat, colour, clarity, certification, price.
# Each row is one diamond. If everything appeared in a single column,
# the separator would be wrong and read_csv would need sep=";".
```

## **3. Basic Dataset Information**

```python
# 3. Basic information about the table
# These four checks are run before any statistic is computed.

print("Dataset shape:", df.shape)
print("\nColumn names:", df.columns.tolist())
print("\nData types:")
print(df.dtypes)

# Count missing values in every column.
print("\nMissing values per column:")
print(df.isnull().sum())

# Interpretation:
# 308 rows and 5 columns. carat is float64, price is int64,
# the three remaining columns hold text: pandas labels them object,
# or str in pandas 3 and later. Either label means the same thing here.
# There are no missing values, so nothing has to be dropped or filled.
# A statistic computed on a table that has not been checked means nothing.
```

## **4. Identify Variable Types**

```python
# 4. Split the columns into numerical and categorical
# The split decides which statistics are meaningful for which column.

numerical_cols = ["carat", "price"]
categorical_cols = ["colour", "clarity", "certification"]

num_df = df[numerical_cols]

print("Numerical columns:", numerical_cols)
print("Categorical columns:", categorical_cols)

# Interpretation:
# Numerical columns get a mean, a median and a standard deviation.
# Categorical columns get counts and a mode.
# The average of GIA, HRD and IGI does not exist.
```

## **5. Descriptive Statistics for Numerical Variables**

```python
# 5.1 Sample size
# The number of observations the statistics below are based on.

n_obs = len(num_df)
print(f"Sample size (number of observations): {n_obs}")

# Interpretation:
# Every statistic in this section describes these 308 stones and nothing else.
```

```python
# 5.2 Minimum and maximum
# The two ends of the observed range.

print("Minimum values:")
print(num_df.min())
print("\nMaximum values:")
print(num_df.max())

# Interpretation:
# carat runs from 0.18 to 1.10, price from 638 to 16008 Singapore dollars.
# The first job of min and max is not description but control:
# a price of 0, a negative weight or a carat of 110 would signal a data error.
```

```python
# 5.3 Mean
# The arithmetic average of each numerical column.

print("Sample mean:")
print(num_df.mean())

# Interpretation:
# Average weight 0.63 carat, average price 5019.48 SGD.
# The mean is the balance point of the data: every observation pulls on it.
# One stone priced at 100000 would visibly move the average of 308 stones.
```

```python
# 5.4 Median
# The middle value once the column is sorted.

print("Median:")
print(num_df.median())

# Compare the two measures of the centre directly.
print("\nMean minus median:")
print(num_df.mean() - num_df.median())

# Interpretation:
# Median weight 0.62 carat, median price 4215 SGD.
# Half the stones cost less than 4215, half cost more.
# The mean price exceeds the median by about 804 SGD, so the distribution
# is skewed to the right: a few expensive stones pull the average up.
# For carat the two measures almost coincide, so that column is nearly symmetric.
```

```python
# 5.5 Variance and standard deviation
# Two measures of spread around the mean.

print("Sample variance:")
print(num_df.var(ddof=1))
print("\nSample standard deviation:")
print(num_df.std(ddof=1))

# Interpretation:
# Standard deviation of price is 3403.12 SGD, of carat 0.28 carat.
# Variance is the average squared deviation from the mean, so it is measured
# in squared units and cannot be compared with the data itself.
# The standard deviation is its square root and is back in the original units.
# ddof=1 divides by n-1 rather than by n. These 308 stones are a sample drawn
# from a much larger market, and dividing by n-1 corrects the systematic
# underestimation that arises when the mean is estimated from the same sample.
```

```python
# 5.6 Quartiles and percentiles
# Positions in the sorted column rather than distances from the mean.

print("Quartiles (25%, 50%, 75%):")
print(num_df.quantile([0.25, 0.50, 0.75]))

print("\n5th and 95th percentiles:")
print(num_df.quantile([0.05, 0.95]))

# Interpretation:
# For price: Q1 = 1625, Q2 = 4215, Q3 = 7446.
# A quarter of the stones cost less than 1625 SGD, three quarters less than 7446.
# Q2 is the median by definition.
# The 5th and 95th percentiles, 919 and 10655.60, describe the tails:
# only one stone in twenty costs more than 10656 SGD.
```

```python
# 5.7 Interquartile range and the 1.5 x IQR rule
# The width of the middle half of the data, and the standard outlier test.

for col in numerical_cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    print(f"{col}: Q1 = {q1}, Q3 = {q3}, IQR = {iqr:.2f}")
    print(f"    bounds: {lower:.2f} to {upper:.2f}, outliers found: {len(outliers)}\n")

# Interpretation:
# The IQR of price is 7446 - 1625 = 5821 SGD: the middle half of the stones
# lie within that band. The IQR ignores the extreme values entirely,
# which is exactly why it is used together with the median.
# The 1.5 x IQR rule puts the upper bound for price at 16177.50, while the most
# expensive stone costs 16008, so this dataset contains no outliers by that rule.
# The same rule draws the whiskers of a box plot, which is the subject of seminar 3.
```

```python
# 5.8 Skewness and kurtosis
# Two numbers describing the shape of the distribution.

print("Skewness:")
print(num_df.skew())
print("\nKurtosis (excess):")
print(num_df.kurtosis())

# Interpretation:
# Skewness measures asymmetry. Positive means a long tail to the right.
# price 0.66: clearly skewed to the right, which agrees with mean > median.
# carat 0.01: practically symmetric.
# Kurtosis measures how heavy the tails are relative to a normal distribution.
# pandas returns EXCESS kurtosis (Fisher's definition): 0 corresponds to a normal
# distribution, not 3. Textbooks and some spreadsheet functions report the value
# without that correction, so the two differ by exactly 3.
# price -0.32 and carat -1.24: both have lighter tails than a normal curve.
```

```python
# 5.9 describe(): the same statistics in one line
# Shown after the manual computation, not instead of it.

print(num_df.describe().round(2))

# Interpretation:
# describe() returns count, mean, standard deviation, minimum, the three
# quartiles and the maximum. Every one of these was computed by hand above.
# In practice this is the first command applied to a new numerical column;
# it is placed at the end of the section so that its output can be read,
# not merely printed.
# Source: https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.describe.html
```

## **6. Descriptive Statistics for Categorical Variables**

```python
# 6.1 Frequency tables
# The count of observations in each category.

for col in categorical_cols:
    freq_table = df[col].value_counts().sort_index()
    print(f"Frequency table for '{col}':")
    print(freq_table)
    print(f"Total: {freq_table.sum()}\n")

# Interpretation:
# A frequency table is the complete description of a categorical column.
# colour: F is the most common with 82 stones, D the rarest with 16.
# certification: GIA 151, HRD 79, IGI 78.
# sort_index() orders the categories alphabetically; without it pandas
# orders them by frequency, which hides gaps in an ordered scale.
```

```python
# 6.2 Relative frequencies
# The same tables expressed as shares of the sample.

for col in categorical_cols:
    shares = (df[col].value_counts(normalize=True).sort_index() * 100).round(1)
    print(f"Shares for '{col}', per cent:")
    print(shares)
    print()

# Interpretation:
# GIA certifies 49.0 per cent of the stones, HRD 25.6 and IGI 25.3.
# Shares are easier to compare than counts and, unlike counts, they allow
# two datasets of different size to be compared directly.
```

```python
# 6.3 Mode
# The most frequent value of a column.

mode_categorical = df[categorical_cols].mode().iloc[0]
print("Mode for categorical variables:")
print(mode_categorical)

# For a numerical column the mode is reported in full, because there may be several.
print("\nAll modes of carat:")
print(df["carat"].mode().tolist())

# Interpretation:
# colour F, clarity VS1, certification GIA. For a categorical column the mode
# is the natural measure of the centre, since no mean exists.
# For a continuous column the mode is weak: repeated values are largely
# accidental. The mode of carat is 1.0 only because round weights are
# preferred in the trade, not because the distribution peaks there.
# .mode() returns every value tied for first place, so .iloc[0] silently
# keeps the first one and should not be used without checking the full list.
```

```python
# 6.4 describe() for categorical columns
# The categorical counterpart of section 5.9.

print(df[categorical_cols].describe())

# Interpretation:
# For object columns describe() reports the number of observations, the number
# of distinct categories, the most frequent category and its frequency.
# colour has 6 categories, clarity 5, certification 3.
```

## **7. Comparing Groups**

```python
# 7.1 Statistics computed separately for each group
# groupby splits the rows, applies the same statistic to each part, and combines the results.

print(df.groupby("certification")["price"].agg(["mean", "median", "count"]).round(0))

# Interpretation:
# Mean price: GIA 5310, HRD 7181, IGI 2267 SGD.
# Stones certified by IGI are on average less than a third as expensive as
# those certified by HRD. A single command turns the question
# "is price related to the laboratory?" into a table that can be read.
```

```python
# 7.2 A group mean can mislead
# The same command applied to colour, which is an ordered scale.

print(df.groupby("colour")["price"].agg(["mean", "median", "count"]).round(0))

# Interpretation:
# Colour runs from D (best) to I (worst), so the mean price would be expected
# to fall from left to right. It does not: D 7100, E 5508, F 4786, G 4426,
# H 4965, I 5175. The worst colours are not the cheapest.
# The reason is that the groups differ in more than colour: they also differ
# in weight, and weight dominates the price. A group mean compares groups
# as they are, not one feature at a time. Isolating the effect of colour
# requires holding carat constant, which is the subject of regression analysis.
```

## **8. Bivariate Analysis: Two Numerical Variables**

```python
# 8.1 Covariance
# A measure of whether two numerical variables move together.

cov_matrix = num_df.cov()
print("Covariance matrix:")
print(cov_matrix)

cov_carat_price = cov_matrix.loc["carat", "price"]
print(f"\nCovariance between carat and price: {cov_carat_price:.2f}")

# Interpretation:
# The covariance is positive, 891.15: when a stone is heavier than average
# it also tends to be more expensive than average.
# A negative covariance would indicate the opposite relationship.
# The diagonal of the matrix holds the variances, because the covariance
# of a variable with itself is its variance.
# The number 891.15 by itself, however, cannot be judged large or small.
```

```python
# 8.2 Why covariance is not enough
# The same data with the price expressed in thousands of dollars.

df_scaled = df.copy()
df_scaled["price_k"] = df_scaled["price"] / 1000

print("Covariance with price in dollars:  ",
      round(df_scaled["carat"].cov(df_scaled["price"]), 2))
print("Covariance with price in thousands:",
      round(df_scaled["carat"].cov(df_scaled["price_k"]), 2))
print("Correlation with price in dollars:  ",
      round(df_scaled["carat"].corr(df_scaled["price"]), 4))
print("Correlation with price in thousands:",
      round(df_scaled["carat"].corr(df_scaled["price_k"]), 4))

# Interpretation:
# Nothing about the stones has changed, only the unit of measurement.
# The covariance falls from 891.15 to 0.89, that is, by a factor of a thousand,
# while the correlation stays at 0.9447.
# Covariance depends on the units of both variables and therefore cannot be
# compared across pairs of variables or across datasets. Correlation is the
# covariance divided by both standard deviations, which removes the units.
```

```python
# 8.3 Correlation
# The standardised measure of a linear relationship.

correlation_carat_price = df["carat"].corr(df["price"])
print(f"Correlation between carat and price: {correlation_carat_price:.4f}")

print("\nCorrelation matrix:")
print(num_df.corr())

# Interpretation:
# 0.9447 indicates a very strong positive linear relationship: heavier stones
# are almost uniformly more expensive.
# The coefficient always lies between -1 and +1, and the diagonal of the
# matrix is always 1.
# Two cautions. A value near 0 means no LINEAR relationship, not no
# relationship: a perfect U-shaped dependence also yields a correlation near 0.
# And correlation is not causation. Here weight plausibly drives price,
# but the coefficient itself is no evidence of that.
```

## **9. Bivariate Analysis: Two Categorical Variables**

```python
# 9.1 Contingency table
# The joint frequency distribution of two categorical columns.

contingency_table = pd.crosstab(df["colour"], df["certification"])
print("Contingency table (colour against certification):")
print(contingency_table)

# The same table as row shares, which is easier to compare.
print("\nRow shares, per cent:")
print((pd.crosstab(df["colour"], df["certification"], normalize="index") * 100).round(1))

# Interpretation:
# The table counts how often each pair of categories occurs together.
# GIA appears across every colour, while the split between HRD and IGI
# varies noticeably from one colour to another.
# If the distribution of certification were the same in every row,
# the two variables would be unrelated. Large differences suggest an
# association worth testing formally, for example with a chi-square test.
```

## **10. Summary**

- The table is checked before anything is computed: `shape`, `dtypes`, `isnull().sum()`, `head()`.
- Centre: the mean is sensitive to extreme values, the median is not. A mean above the median indicates a tail to the right.
- Spread: variance and standard deviation measure distance from the mean; quartiles and the interquartile range measure position in the sorted column.
- Shape: skewness measures asymmetry, kurtosis measures the weight of the tails. pandas reports excess kurtosis, where 0 corresponds to a normal distribution.
- Categorical columns are described by frequencies, shares and the mode; no mean exists for them.
- `groupby` compares groups, but a group mean reflects every difference between the groups, not only the one of interest.
- Covariance depends on the units of measurement; correlation does not. Correlation captures a linear relationship only and is not evidence of causation.
- `describe()` reproduces most of section 5 in one line and is the natural starting point once its output can be read.
